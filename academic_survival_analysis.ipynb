# Academic Survival Longitudinal Analysis
Python EDA of academic, engagement and financial factors associated with dropout events.

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('../data/academic_survival_cleaned.csv')
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
df.groupby('Target_Dropout_Next_Sem')[['Sem_GPA','Attendance','LMS_Logins','Work_Hours','Financial_Stress','Failed_Courses']].mean().round(2)
```

```python
(df.groupby('Semester')['Target_Dropout_Next_Sem'].mean()*100).plot(marker='o')
plt.ylabel('Dropout event rate (%)')
plt.title('Dropout Event Rate by Semester')
plt.show()
```
